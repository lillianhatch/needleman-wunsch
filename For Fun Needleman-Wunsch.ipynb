```python
#This program runs a Needleman-Wunsch Algorithm
#import required libraries
import numpy as np
```

```python
#All required user input data is gathered
print("This program runs a Needleman-Wunsch algorithm. Please provide the following parameters")
seq1 = input("What is the first sequence: ")
seq2 = input("What is the second sequence: ")
match_score = float(input("What is the match score: "))
mis_penalty = float(input("What is the mismatch penalty: "))
gap_penalty = float(input("What is the gap penalty: "))

#check penalties are less than the match score
if match_score < 0:
  print("Match score is less than 0, are you sure?")
if gap_penalty > match_score:
  print("Gap penalty is greater than match score, are you sure?")
if gap_penalty > 0:
  print("Gap penalty is greater than 0, are you sure?")
if mis_penalty > match_score:
  print("Mismatch penalty is greater than match score, are you sure?")
if mis_penalty > 0:
  print("Mismatch penalty is greater than 0, are you sure?")
```

```python
#Similarity function to determine if sequence letters are the same

def simFx(a, b, match_score, mis_penalty):
  if a == b:
    score = match_score
  else:
    score = mis_penalty
  return score
```

```python
#Initialize the matrix for sequence comparison
matrix = np.zeros((len(seq1)+1, len(seq2)+1))

#Initialize the gap penalties in the first row
for i in range(1, len(seq2)+1):
  matrix[0][i] = matrix[0][i-1] + gap_penalty

#Initialize the gap penalties in the first column
for j in range(1, len(seq1)+1):
  matrix[j][0] = matrix[j-1][0] + gap_penalty
```

```python
#Fill the rest of the matrix with the max scores
for i in range(1, len(seq1)+1):
  for j in range(1, len(seq2)+1):
    matrix[i][j] = (max(matrix[i-1][j-1] + simFx(seq1[i-1],
        seq2[j-1], match_score, mis_penalty),
        matrix[i-1][j] + gap_penalty,
        matrix[i][j-1] + gap_penalty))

seqAlignScore = matrix[len(seq1)][len(seq2)]
```

```python
#traceback loop
#we have to figure out where each cell came from and record the path

i = len(seq1)
j = len(seq2)
seq1Aligned = ""
seq2Aligned = ""

while i>0 or j>0:
    if matrix[i][j] == matrix[i-1][j-1] + simFx(seq1[i-1],
        seq2[j-1], match_score, mis_penalty):
        seq1Aligned = seq1[i-1] + seq1Aligned
        seq2Aligned = seq2[j-1] + seq2Aligned
        i = i-1
        j = j-1

    elif matrix[i][j] == matrix[i-1][j] + gap_penalty:
        seq1Aligned = seq1[i-1] + seq1Aligned
        seq2Aligned = "-" + seq2Aligned
        i = i-1

    elif matrix[i][j] == matrix[i][j-1] + gap_penalty:
        seq1Aligned = "-" + seq1Aligned
        seq2Aligned = seq2[j-1] + seq2Aligned
        j = j-1
```

```python
#printing all data

print(matrix, "\n")
print("The sequence's alignment score is: ", seqAlignScore)
print(seq1Aligned, "\n")
print(seq2Aligned)
```
